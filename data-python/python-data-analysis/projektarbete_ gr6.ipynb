```python
import matplotlib
import matplotlib.pyplot as plt
%matplotlib inline
matplotlib.style.use('ggplot')

plt.scatter(data.AnnualIncome , data.CustomerID)
plt.ylabel('CustomerID')
plt.xlabel('AnnualIncome')
plt.title('Correlation between CustomerID and AnnualIncome')
plt.show()
```

We have information on 2000 customers and their yearly earnings are around $180,000.

```python
plt.scatter(data['Work Experience'], data.AnnualIncome)
plt.ylabel('Work Experience')
plt.xlabel('AnnualIncome')
plt.title('Correlation between Work Experience and AnnualIncome')
plt.show()
```

The max years of  work experience that customers have are over 17 years and the max annual income is $180,000.
